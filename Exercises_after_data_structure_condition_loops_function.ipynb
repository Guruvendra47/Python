#Solution

```python
# Netflix Customer Analytics System

customers = []


# 1. Add Customer
def add_customer():
    name = input("Enter customer name: ")
    country = input("Enter country: ")
    subscription = input("Enter subscription (Basic/Standard/Premium): ")
    watch_hours = int(input("Enter watch hours: "))

    customer = {
        "name": name,
        "country": country,
        "subscription": subscription,
        "watch_hours": watch_hours
    }

    customers.append(customer)

    print("Customer added successfully!")


# 2. View All Customers
def view_customers():

    if len(customers) == 0:
        print("No customers available.")
        return

    print("\n--- Customer Records ---")

    for customer in customers:
        print(customer)



# 3. Find Heavy Watchers
def heavy_watchers():

    print("\n--- Heavy Watchers (>20 hours) ---")

    found = False

    for customer in customers:

        if customer["watch_hours"] > 20:
            print(customer["name"], "-", customer["watch_hours"], "hours")
            found = True

    if found == False:
        print("No heavy watchers found.")



# 4. Find Top Viewer
def top_viewer():

    if len(customers) == 0:
        print("No customers available.")
        return

    top_customer = customers[0]

    for customer in customers:

        if customer["watch_hours"] > top_customer["watch_hours"]:
            top_customer = customer


    print("\n--- Top Viewer ---")
    print(top_customer)



# 5. Calculate Average Watch Hours
def average_watch_hours():

    if len(customers) == 0:
        print("No data available.")
        return


    total_hours = 0

    for customer in customers:
        total_hours += customer["watch_hours"]


    average = total_hours / len(customers)

    print("Average Watch Hours:", average)



# 6. Subscription Report
def subscription_report():

    report = {}

    for customer in customers:

        subscription = customer["subscription"]

        if subscription in report:
            report[subscription] += 1

        else:
            report[subscription] = 1


    print("\n--- Subscription Report ---")

    print(report)



# 7. Upgrade Subscription
def upgrade_subscription():

    name = input("Enter customer name to upgrade: ")


    for customer in customers:

        if customer["name"].lower() == name.lower():

            current = customer["subscription"]


            if current == "Basic":
                customer["subscription"] = "Standard"

            elif current == "Standard":
                customer["subscription"] = "Premium"

            else:
                print("Already Premium subscription.")
                return


            print("Subscription upgraded!")
            print(customer)

            return


    print("Customer not found.")



# 8. Search Customer by Country
def search_by_country():

    country = input("Enter country: ")

    found = False


    print("\n--- Customers from", country, "---")


    for customer in customers:

        if customer["country"].lower() == country.lower():

            print(customer)
            found = True


    if found == False:
        print("No customers found.")



# ---------------- MAIN MENU ----------------


while True:

    print("\n====== Netflix Customer Analytics ======")

    print("1. Add Customer")
    print("2. View All Customers")
    print("3. Find Heavy Watchers")
    print("4. Find Top Viewer")
    print("5. Calculate Average Watch Hours")
    print("6. Subscription Report")
    print("7. Upgrade Subscription")
    print("8. Search Customer by Country")
    print("9. Exit")


    choice = input("Enter your choice: ")



    if choice == "1":
        add_customer()

    elif choice == "2":
        view_customers()

    elif choice == "3":
        heavy_watchers()

    elif choice == "4":
        top_viewer()

    elif choice == "5":
        average_watch_hours()

    elif choice == "6":
        subscription_report()

    elif choice == "7":
        upgrade_subscription()

    elif choice == "8":
        search_by_country()

    elif choice == "9":
        print("Exiting Netflix System...")
        break

    else:
        print("Invalid choice. Try again.")
```
